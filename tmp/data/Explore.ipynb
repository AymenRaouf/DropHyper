```python
from data import *
data = "cocitation" 
dataset = "pubmed"

def load(data, dataset, split = 1):
    """
    parses the dataset
    """
    parsed_dataset = parser(data, dataset).parse()

    Dir, _ = "./"
    
    file = os.path.join(Dir, data, dataset, "splits", str(split) + ".pickle")
    if not os.path.isfile(file): print("split + ", str(split), "does not exist")
    with open(file, 'rb') as H: 
        Splits = pickle.load(H)
        train, test = Splits['train'], Splits['test']

    return parsed_dataset, train, test
X, Y, G = load(data, dataset)
```
